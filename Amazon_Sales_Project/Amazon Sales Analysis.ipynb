```python
def filter_columns(df,column_name,dtype,skip_element = None,show_str_value = False,show_unwanted_elements=False):
    column = df[column_name]
    unwanted_symbols = []
    for i in column:
        i = str(i)
        for j in i:
            if type(skip_element) == list:
                if j in skip_element:
                    continue
            else:
                if j == skip_element:
                    continue
            try:
                j = int(j)
            except:
                if j not in unwanted_symbols:
                    unwanted_symbols.append(j)

#     print(unwanted_symbols)
    
    
    for i in unwanted_symbols:
        
        column = column.str.replace(i,'')
    
    if show_str_value == True:
        for i in column:
            try:
                float(i)
            except:
                print(i)
    
    if show_unwanted_elements == True:
        print(unwanted_symbols)
    
    return column.astype(dtype) 
```

```python
df['discounted_price'] = filter_columns(df,'discounted_price',float,skip_element = '.')
```

```python
df['actual_price'] = filter_columns(df,'actual_price','float',skip_element = '.')
```

```python
df['discount_percentage'] = filter_columns(df,'discount_percentage','float',skip_element = '.')
```

```python
df['rating'] = filter_columns(df,'rating','float',skip_element = ['.'],show_str_value=True,show_unwanted_elements=True)
```

###### This error is shows us because of this ['|'] symbol

```python
df[df['rating'] == '|']
```

```python
# We remove the row which contain "|" or sign. Because we want to convert our rating column into my float datatype

df.drop(df[df['rating'] == '|'].index[0],axis = 0,inplace= True)
```

```python
df['rating'] = filter_columns(df,'rating','float')
```

```python
df['rating_count'] = filter_columns(df,'rating_count','float')
```

### What are the top rating_count products by category?

### How do different percentage discounts contribute to the overall distribution of discounted prices?

```python
df[['discounted_price','actual_price','discount_percentage']].describe()
```

```python
labels = ['0 - 20','21 - 40','41 - 60','61 - 80','81 - 95']
bins = [-1,20,40,60,80,95]
df['discounted_range'] = pd.cut(df['discount_percentage'],bins = bins , labels = labels)
```

(-1, 20] → '0 - 20'

(20, 40] → '21 - 40'

(40, 60] → '41 - 60'

(60, 80] → '61 - 80'

(80, 95] → '81 - 95'

```python
discount_range = (round(df['discounted_range'].value_counts(normalize = True) * 100,2)).reset_index()
discount_range.rename(columns = {"index" : 'Discounted_range',"discounted_range" : "Percentage"},inplace = True)
```

```python
discount_range
```

```python
plt.pie(discount_range['Percentage'],labels = discount_range['Discounted_range'],autopct = "%0.2f%%")
print()
plt.title('Partition of Discounted Prices According to Percentage Share')
plt.show()
```

### What is the distribution of discounted prices vs. actual prices?

```python
sns.histplot(df['actual_price'], color='blue', kde=True, label='Actual Price')
sns.histplot(df['discounted_price'], color='red', kde=True, label='Discounted Price')
plt.title('Distribution of Actual Prices vs. Discounted Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
